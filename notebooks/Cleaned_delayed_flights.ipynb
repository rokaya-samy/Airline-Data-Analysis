```python
#Renaming columns for better understanding
Delayed_Flights.rename(columns={'Year': 'year','Month': 'month','DayofMonth': 'day','DepTime': 'Actual_DepTime','CRSDepTime': 'Scheduled_DepTime','ArrTime':'Actual_ArrTime','CRSArrTime':'Scheduled_ArrTime','CRSElapsedTime':'Expected_ElapsedTime','ArrDelay':'Arrival_Delay','DepDelay':'Departure_Delay',}, inplace=True)
Delayed_Flights.head(10)
```

```python
# Creating a new column for the date of the flight
Delayed_Flights['Date']=pd.to_datetime(Delayed_Flights[['year', 'month', 'day']])
Delayed_Flights.head(10)
```

```python
# List of columns to be processed
time_cols = ['Actual_DepTime', 'Scheduled_DepTime', 'Actual_ArrTime', 'Scheduled_ArrTime']

# Perform conversion in a single block
for col in time_cols:
    # Handle NaNs, convert to float (to manage dots), then int, then string with zfill
    Delayed_Flights[col] = Delayed_Flights[col].fillna(0).astype(float).astype(int).astype(str).str.zfill(4)

# Verification
print(Delayed_Flights[time_cols].dtypes)
Delayed_Flights[time_cols].head()
Delayed_Flights.info()
```

```python
# Replace invalid value (24:00) with 00:00 
time_cols = ['Scheduled_DepTime', 'Actual_DepTime', 'Scheduled_ArrTime', 'Actual_ArrTime']
Delayed_Flights[time_cols] = Delayed_Flights[time_cols].replace('2400', '0000')
Delayed_Flights[time_cols].head(10)
```

```python

time_cols = ['Scheduled_DepTime', 'Actual_DepTime', 'Scheduled_ArrTime', 'Actual_ArrTime']
for col in time_cols:
    Delayed_Flights[col] = Delayed_Flights[col].astype(str).str.replace('.0', '', regex=False).str.zfill(4)
    Delayed_Flights[col] = pd.to_datetime(Delayed_Flights[col], format='%H%M', errors='coerce').dt.time
Delayed_Flights[time_cols].head(10)
```

```python
# Handlling Null Values 

Delayed_Flights.fillna({'CarrierDelay': 0, 'WeatherDelay': 0, 'NASDelay': 0, 'SecurityDelay': 0, 'LateAircraftDelay': 0}, inplace=True)
Delayed_Flights.dropna(subset=['TailNum'], inplace=True)
Delayed_Flights.isnull().sum()
```

```python
Delayed_Flights['Delay_sum'] = (
    Delayed_Flights['CarrierDelay'] +
    Delayed_Flights['WeatherDelay'] +
    Delayed_Flights['NASDelay'] +
    Delayed_Flights['SecurityDelay'] +
    Delayed_Flights['LateAircraftDelay']
)
```

```python
initial_rows = len(Delayed_Flights)
Delayed_Flights.drop_duplicates(inplace=True)
final_rows = len(Delayed_Flights)

print(f"Removed {initial_rows - final_rows} duplicate rows.")
print(f"Remaining rows: {final_rows}")
```

```python
Delayed_Flights
```

```python
Delayed_Flights['Year'] = Delayed_Flights['Date'].dt.year
Delayed_Flights['Month'] = Delayed_Flights['Date'].dt.month
Delayed_Flights['Day'] = Delayed_Flights['Date'].dt.day

Delayed_Flights['Month_Name'] = Delayed_Flights['Date'].dt.month_name()
Delayed_Flights['Day_Name'] = Delayed_Flights['Date'].dt.day_name()

Delayed_Flights['DayOfWeek_Egypt'] = (Delayed_Flights['Date'].dt.dayofweek + 2) % 7 + 1

Delayed_Flights['Month_Name'] = Delayed_Flights['Month_Name'].astype('category')
Delayed_Flights['Day_Name'] = Delayed_Flights['Day_Name'].astype('category')

Delayed_Flights[['Date', 'Year', 'Month_Name', 'Day_Name', 'DayOfWeek_Egypt']].head()
```

```python
Delayed_Flights['Month'] = Delayed_Flights['Date'].dt.month
```

```python
final_order = [
    'Date',              
    'Year',               
    'Month_Name', 
    'Month',       
    'Day',               
    'Day_Name',         
    'DayOfWeek_Egypt',    
    
    'UniqueCarrier', 'FlightNum', 'TailNum',
    'Origin', 'Dest', 'Distance',
    
    'Scheduled_DepTime', 'Actual_DepTime', 'Departure_Delay',
    'Scheduled_ArrTime', 'Actual_ArrTime', 'Arrival_Delay',
    
    'CarrierDelay', 'WeatherDelay', 'NASDelay', 'SecurityDelay', 'LateAircraftDelay', 'Delay_sum',
    
    'ActualElapsedTime', 'Expected_ElapsedTime', 'AirTime', 'TaxiIn', 'TaxiOut',
    
    'Cancelled', 'CancellationCode', 'Diverted'
]
Delayed_Flights = Delayed_Flights[final_order]
Delayed_Flights.head()
```
